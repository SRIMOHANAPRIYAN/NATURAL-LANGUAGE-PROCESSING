# LOAD DATASET

```python
import re
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

with open('amjath.txt', 'r', encoding='utf-8') as f:
    raw_text = f.readlines()
```

# PREPROCESS THE DATA FOR EXTRACTING MEANINGFULL MESSAGE

```python
messages = []
pattern = re.compile(r'\d{1,2}/\d{1,2}/\d{2,4}\s+\d{1,2}:\d{2}\s*(AM|PM)?\s*[\w]+\s*(.*)')
for line in raw_text:
    if '<Media omitted>' in line or not line.strip():
        continue
    split_line = line.split('\t')
    if len(split_line) >= 4:
        message = split_line[3].strip()
        message = re.sub(r'http[s]?://\S+', '', message)
        if message and '<Media omitted>' not in message:
            messages.append(message)

corpus = [msg for msg in messages if msg and not msg.isspace()]
```

# TF-IDF VECTORIZATION

```python
vectorizer = TfidfVectorizer(stop_words='english',max_features=50)
tfidf_matrix = vectorizer.fit_transform(corpus)
```

# SHOW TOP FEATURES FOR THE WHOLE FILES

```python
feature_array = vectorizer.get_feature_names_out()
tfidf_scores = tfidf_matrix.sum(axis=0).A1
tfidf_ranked = sorted(zip(feature_array, tfidf_scores), key=lambda x: x[1], reverse=True)
```

```python
print("Top TF-IDF words:")
for word, score in tfidf_ranked:
    print(f"{word}: {score:.3f}")
```
