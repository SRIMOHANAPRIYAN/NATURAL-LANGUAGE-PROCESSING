```python
import re 
from sklearn.feature_extraction.text import HashingVectorizer
with open('amjath.txt', 'r', encoding='utf-8') as f:
    raw_text = f.readlines()
```

# PREPROCESSING THE DATA TO EXTRACT MEANINGFULL DATA

```python
messages = []
for line in raw_text:
    split_line = line.split('\t')
    if len(split_line) >= 4:
        message = split_line[3].strip()
        message = re.sub(r'http[s]?://\S+', '', message)
        if message and '<Media omitted>' not in message:
            messages.append(message)
corpus = [msg for msg in messages if msg and not msg.isspace()]
```

# HASH VECTORIZATION

```python
vectorizer = HashingVectorizer(n_features=50,stop_words='english',alternate_sign=False)
hash_vectors = vectorizer.transform(corpus)
```

# SHAPE AND SAMPLE VECTOR

```python
print("Hash vector shape:", hash_vectors.shape)
print("First message hash vector:\n", hash_vectors[0].toarray())
```

# PRINT HASH VECTOR FOR FEW SAMPLE MESSAGES

```python
for i, vec in enumerate(hash_vectors[:3]):
    print("Hash vector for message {i+1}:",vec.toarray())
```
